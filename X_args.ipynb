```python
def multiply(*numbers):
  total = 1
  for number in numbers:
    total *= number
  return total

print(multiply (2, 3, 4, 5))
```

```python
def subtract(*numbers):
  total = 0
  for number in numbers:
    total -= number
  return total

print(subtract(10, 6, 7, 8, 4, 2))
```
